```python
import subprocess

entrada = "beer-mobile.webp"
salida = "beer-mobile.webm"

comando = [
    "ffmpeg",
    "-i", entrada,
    "-vf", "scale=180:-2,fps=12",
    "-an",
    "-c:v", "libvpx-vp9",
    "-crf", "35",
    "-b:v", "0",
    salida
]

subprocess.run(comando, check=True)

print("Conversión completada:", salida)
```
